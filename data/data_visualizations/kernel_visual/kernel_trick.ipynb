```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D
```

```python
np.random.seed(99)
#generate some sample data in 2d, with 2 classes as concentric circles, one surrounding the other
n_points = 100
radius_inner = 3
radius_outer = 4
theta = np.linspace(0, 2*np.pi, n_points)
class_1 = np.array([[radius_inner*np.cos(t)+ np.random.randn()*0.1,radius_inner*np.sin(t)+ np.random.randn()*0.1] for t in theta])
class_2 = np.array([[radius_outer*np.cos(t)+np.random.randn()*0.1, radius_outer*np.sin(t)+np.random.randn()*0.1] for t in theta])

#plot the sample data 
plt.figure(figsize=(6,6))
plt.scatter(class_1[:,0],class_1[:,1], color='red', label='class 1')
plt.scatter(class_2[:,0], class_2[:,1], color='blue', label='class 2')
plt.legend()
plt.show()

#project the data into 3d by adding a 3rd term of the first two squared
class_1_3d = np.array([[x, y, x**2 +y**2] for x, y in class_1])
class_2_3d = np.array([[x, y, x**2+y**2] for x, y in class_2])

#plot the new vectors in 3d 
fig = plt.figure(figsize=(6,6))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(class_1_3d[:,0], class_1_3d[:,1], class_1_3d[:,2], color='red', label='class 1')
ax.scatter(class_2_3d[:,0], class_2_3d[:,1], class_2_3d[:,2], color='blue', label='class 2')
ax.legend()
plt.show()
```
